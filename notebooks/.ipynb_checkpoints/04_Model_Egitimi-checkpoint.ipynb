```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score

df_final = pd.read_csv('C:/labs/data/islenmis/perfect_dataset_for_ml.csv')


#1. Kritik adım : Kategorik kolonları yapay zekaya doğru tanıtma
#Sayısal büyüklük tuzağına düşmmeesi için kolonları kategorik stringe çeviriyoruz
categorical_cols = ['ip.proto', 'tcp_flags' , 'src_port_category' , 'dst_port_category' , 'packet_size_category']
for col in categorical_cols:
    if col in df_final.columns:
        df_final[col] = df_final[col].astype(str)

#2. One Hot Encoding 
X = df_final.drop(columns=['label'])
X = pd.get_dummies(X , drop_first=True)
y = df_final['label']

# Eğitime girmeyecek, hedef sızıntısı ve araç tabanlı ezber yaratan kolonları eliyoruz
# tcp.window_size_value kolonunu tamamen uçuruyoruz
X = df_final.drop(columns=['label', 'tcp.window_size_value'], errors='ignore')

# Kategorik kolonları dummy (One-Hot) formatına açıyoruz
X = pd.get_dummies(X, drop_first=True)
y = df_final['label']

# Tekrar Eğitim ve Test Ayrımı
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"[+] Yeni eğitilecek öznitelikler: {X.columns.tolist()}")

# Modelin Aşırı Derine İnip Ezberlemesini Önlemek İçin Derinliği (max_depth) Biraz Daha Sıkıştırıyoruz
rf_model = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# Yeni tahminler ve rapor
y_pred = rf_model.predict(X_test)
print("\n================ YENİ RAFİNE PERFORMANS RAPORU ================")
print(f"Genel Doğruluk (Accuracy) Skoru: {accuracy_score(y_test, y_pred):.4f}")
print(classification_report(y_test, y_pred, target_names=['Normal (0)', 'Saldırı (1)']))

X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.3, random_state=42 , stratify=y)

#Random forest modelinin sınırlarının belirleyerek eğitilmesi
## n_jobs=-1 kullanarak bilgisayarındaki tüm işlemci çekirdeklerini tam performans çalıştırıyoruz
print("\n[*] Random Forest Modeli Eğitiliyor")
rf_model = RandomForestClassifier(n_estimators=100 , max_depth=12 , random_state=42 , n_jobs=-1)
rf_model.fit(X_train ,y_train)
print("Model eğitimi başarıyla tamamlandı")

y_pred = rf_model.predict(X_test)

#Akademik Metriklerin Raporlanması
print("\n================ PERFORMANS RAPORU ================")
print(f"Genel Doğruluk (Accuracy) Skoru: {accuracy_score(y_test, y_pred):.4f}")
print("\nSınıflandırma Detayları:")
print(classification_report(y_test, y_pred, target_names=['Normal (0)', 'Saldırı (1)']))

#Görsel Karmaşıklık Matrisi (Confusion Matrix)
plt.figure(figsize=(6,5))
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Normal', 'Saldırı'], yticklabels=['Normal', 'Saldırı'])
plt.title('Nihai Model - Karmaşıklık Matrisi')
plt.ylabel('Gerçek Sınıf')
plt.xlabel('Yapay Zekanın Tahmini')
plt.tight_layout()
plt.show()
```
