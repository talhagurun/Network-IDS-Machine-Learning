```python
import os
import pandas as pd
from sklearn.utils import resample

#Klasör yollarını tanımla
HAM_VERI_DIZINI = "C:/labs/data/ham"
ISLENMIS_VERI_DIZINI = "C:/labs/data/islenmis"

print("Ham veri setleri yükleniyor...")
df_normal = pd.read_csv(f"{HAM_VERI_DIZINI}/normal.csv")
df_attack = pd.read_csv(f"{HAM_VERI_DIZINI}/attack.csv")
#Eksik satırlar varsa veri setini bozmaması için 0 ile doldur
df_normal.dropna(how='all' , inplace=True)
df_attack.dropna(how='all' , inplace=True)

#Etiketleme
df_normal['label'] = 0
df_attack['label'] = 1

print("Eksikler gittikten sonraki güvenli havuz boyutları")
print(f"Temiz normal paket havuzu : {len(df_normal)} satır")
print(f"Temiz saldırı paket havuzu : {len(df_attack)} satır")

#Rastgele ve karışık şekilde 60K veri çek
target_samples = 60000
print(f"Temizlenmiş havuzlardan rastgele {target_samples} adet benzersiz paket seçiliyor")
df_normal_balanced = resample(df_normal , replace=False , n_samples=target_samples, random_state=42)
df_attack_balanced = resample(df_attack , replace=False , n_samples=target_samples, random_state=42)

#Birleştirme ve Tamamen Karıştırma
gold_df = pd.concat([df_normal_balanced, df_attack_balanced])
gold_df = gold_df.sample(frac=1 , random_state=42).reset_index(drop=True)

print("\n===YENİ DENGELİ VERİ SETİ DAĞILIMI===")
print(gold_df['label'].value_counts())

output_path = f"{ISLENMIS_VERI_DIZINI}/gold_dataset_120k.csv"
gold_df.to_csv(output_path, index = False)
print(f"120.000 satırlık altın veri seti kaydedildi: {output_path}")
```
